```python
df = data
stats = {}
for col in df.columns:
    stats[col] = {
        "Объем выборки": df[col].count(),
        "Дисперсия": df[col].var(),
        "Среднее": df[col].mean(),
        "Медиана": df[col].median(),
        "Нижняя квартиль (Q1)": df[col].quantile(0.25),
        "Верхняя квартиль (Q3)": df[col].quantile(0.75),
        "Мода": df[col].mode().iloc[0] if not df[col].mode().empty else "Нет",
        "Асимметрия": df[col].skew(),
        "Эксцесс": df[col].kurt(),
        "Минимум": df[col].min(),
        "Максимум": df[col].max(),
        "Размах": df[col].max() - df[col].min(),
    }

```

```python
print(pd.DataFrame(stats).round(4))
```
